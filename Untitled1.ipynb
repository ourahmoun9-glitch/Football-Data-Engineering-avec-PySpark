```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, when

# 1. Initialiser la Spark Session
spark = SparkSession.builder \
    .appName("Football_Preprocessing") \
    .getOrCreate()

# 2. Charger le fichier CSV
df = spark.read.csv("/content/drive/MyDrive/Colab Notebooks/data/ensemble-de-donnees-6a0a3f2316332589898898.csv", header=True, inferSchema=True)

# 3. Comprendre la signification des colonnes (Afficher le schéma)
print("--- Structure initiale des données ---")
df.printSchema()

# 4. Suppression de la colonne inutile (Match_ID)
df_prepared = df.drop("Match_ID")

# Afficher un aperçu
df_prepared.show(5)
```

```python
# Création des colonnes indicatrices avec la fonction 'when'
df_with_indicators = df_prepared \
    .withColumn("HomeTeamWin", when(col("FTR") == "H", 1).otherwise(0)) \
    .withColumn("AwayTeamWin", when(col("FTR") == "A", 1).otherwise(0)) \
    .withColumn("GameTie", when(col("FTR") == "D", 1).otherwise(0))

# Vérifier les résultats (Affichage des colonnes créées pour contrôler)
df_with_indicators.show(10)
```

```python
# 1. Filter data to keep only Bundesliga (Div = D1) and seasons between 2000 and 2015
df_filtered = df_with_indicators.filter(
    (col("Div") == "D1") &
    (col("Season") >= 2000) &
    (col("Season") <= 2015)
)

# 2. Check the remaining number of rows after filtering
print(f"Total matches after filtering: {df_filtered.count()}")

# 3. Show a preview of the filtered data
df_filtered.show(5)
```

```python
from pyspark.sql.functions import sum as _sum, count

# 1. Group by BOTH Season and HomeTeam to keep the Season column alive
df_home_seasonal = df_filtered.groupBy("Season", "HomeTeam").agg(
    count("HomeTeam").alias("HomeGamesPlayed"),
    _sum("FTHG").alias("TotalHomeGoalsScored"),
    _sum("FTAG").alias("TotalHomeGoalsConceded"),
    _sum("HomeTeamWin").alias("TotalHomeWins"),
    _sum("GameTie").alias("TotalHomeDraws")
).withColumnRenamed("HomeTeam", "Team")

# 2. Group by BOTH Season and AwayTeam
df_away_seasonal = df_filtered.groupBy("Season", "AwayTeam").agg(
    count("AwayTeam").alias("AwayGamesPlayed"),
    _sum("FTAG").alias("TotalAwayGoalsScored"),
    _sum("FTHG").alias("TotalAwayGoalsConceded"),
    _sum("AwayTeamWin").alias("TotalAwayWins"),
    _sum("GameTie").alias("TotalAwayDraws")
).withColumnRenamed("AwayTeam", "Team")
```

```python
# 1. Join df_home_matches and df_away_matches on the common column 'Team'
# We use an 'inner' join because every team plays both home and away matches
df_merged = df_home_seasonal.join(df_away_seasonal, on="Team", how="inner")

# 2. Verify the structure and see the combined columns
print("--- Merged DataFrame Schema ---")
df_merged.printSchema()

# 3. Show a preview of the merged data for the first 5 teams
df_merged.show(5)
```

```python
from pyspark.sql.functions import col, round

# 1. Join on BOTH Team and Season so Season remains in the final DataFrame
df_merged = df_home_seasonal.join(df_away_seasonal, on=["Team", "Season"], how="inner")

# 2. Calculate basic and advanced metrics
df_totals = df_merged \
    .withColumn("GoalsScored", col("TotalHomeGoalsScored") + col("TotalAwayGoalsScored")) \
    .withColumn("GoalsAgainst", col("TotalHomeGoalsConceded") + col("TotalAwayGoalsConceded")) \
    .withColumn("Win", col("TotalHomeWins") + col("TotalAwayWins")) \
    .withColumn("Tie", col("TotalHomeDraws") + col("TotalAwayDraws")) \
    .withColumn("TotalGames", col("HomeGamesPlayed") + col("AwayGamesPlayed"))

df_totals = df_totals.withColumn("Loss", col("TotalGames") - (col("Win") + col("Tie")))

df_processed = df_totals \
    .withColumn("GoalDifferentials", col("GoalsScored") - col("GoalsAgainst")) \
    .withColumn("WinPercentage", round((col("Win") / col("TotalGames")) * 100, 2)) \
    .withColumn("GoalsPerGame", round(col("GoalsScored") / col("TotalGames"), 2)) \
    .withColumn("GoalsAgainstPerGame", round(col("GoalsAgainst") / col("TotalGames"), 2))
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import col, rank

# 1. Now 'Season' exists in df_processed, so windowSpec will work perfectly!
windowSpec = Window.partitionBy("Season").orderBy(
    col("WinPercentage").desc(),
    col("GoalDifferentials").desc()
)

# 2. Add TeamPosition
football_ranked = df_processed.withColumn(
    "TeamPosition",
    rank().over(windowSpec)
)

# 3. Show the final clean ranking results
print("--- Final Ranking with TeamPosition ---")
football_ranked.select("Season", "TeamPosition", "Team", "WinPercentage", "GoalDifferentials") \
    .orderBy("Season", "TeamPosition").filter(col("Season") == 2012).show(10)
```

```python
from pyspark.sql.functions import col

# ==============================
# Extraction des champions
# ==============================
football_top_teams = football_ranked.filter(
    col("TeamPosition") == 1
)

football_top_teams.show()

# ==============================
# Sauvegarde au format Parquet
# ==============================
# ✅ CORRECTION : sans partitionBy pour garder la colonne Season
football_ranked.write \
    .mode("overwrite") \
    .parquet("/content/football_stats_all")

# ✅ Champions avec Season incluse
football_top_teams.write \
    .mode("overwrite") \
    .parquet("/content/football_top_teams")

print("✅ Les fichiers Parquet ont été sauvegardés avec succès.")
```

```python
# ==============================
# Visualisation des champions
# ==============================
import matplotlib.pyplot as plt

# Convertir Spark DataFrame -> Pandas
champions_pd = football_top_teams.toPandas()

# ==============================
# % de victoires des champions
# ==============================
plt.figure(figsize=(12,5))

plt.bar(
    champions_pd["Season"].astype(str),
    champions_pd["WinPercentage"]
)

plt.title("Pourcentage de victoires des champions")
plt.xlabel("Saison")
plt.ylabel("Win Percentage")

plt.xticks(rotation=45)

plt.show()


# ==============================
# Nombre de buts marqués
# ==============================
plt.figure(figsize=(12,5))

plt.plot(
    champions_pd["Season"].astype(str),
    champions_pd["GoalsScored"],
    marker='o'
)

plt.title("Nombre de buts marqués par les champions")
plt.xlabel("Saison")
plt.ylabel("Goals Scored")

plt.xticks(rotation=45)

plt.show()


# ==============================
# GoalDifferentials par saison
# ==============================
plt.figure(figsize=(12,5))

plt.bar(
    champions_pd["Season"].astype(str),
    champions_pd["GoalDifferentials"]
)

plt.title("Goal Differentials des champions")
plt.xlabel("Saison")
plt.ylabel("Goal Differentials")

plt.xticks(rotation=45)

plt.show()
```
